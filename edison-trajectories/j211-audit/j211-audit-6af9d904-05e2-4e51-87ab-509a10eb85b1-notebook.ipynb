```python
# Numerically derive exact idealized corners/gains and compare explicit SAE coefficients with scipy butter.
import numpy as np, pandas as pd
from scipy import signal
SQRT2=np.sqrt(2)
def sae_coeffs(cfc,fs):
    wa=np.tan(np.pi*2.0775*cfc/fs)
    den=1+SQRT2*wa+wa*wa
    a0=wa*wa/den
    b1=-2*(wa*wa-1)/den
    b2=(-1+SQRT2*wa-wa*wa)/den
    return np.array([a0,2*a0,a0]),np.array([1,-b1,-b2])
def gain(cfc,fs,f):
    b,a=sae_coeffs(cfc,fs)
    _,h=signal.freqz(b,a,worN=2*np.pi*np.asarray(f)/fs)
    return abs(h)**2
fs=1.25e6
freqs=np.array([500.,550.,600.,800.])
rows=[]
for cfc in [180,1000]:
    g=gain(cfc,fs,freqs)
    for f,gg in zip(freqs,g): rows.append([cfc,f,gg,1/gg,20*np.log10(gg)])
print(pd.DataFrame(rows,columns=['CFC','Hz','amplitude_gain','attenuation_x','dB']).to_string(index=False,float_format=lambda x:f'{x:.6g}'))
r=(np.sqrt(2)-1)**0.25
print('two-pass ratio to single-pass corner',r,'pair corner / CFC',r*2.0775)
# exact numerical corners and effective classes based on SAE coefficient definition, not presumed 1.65
for cfc,repo_fc in [(60,100),(180,300),(600,1000),(1000,1650)]:
    correct=2.0775*cfc*r # high fs approx
    repo=repo_fc*r
    print(cfc, correct,repo,'narrow%',100*(repo/correct-1),'effective-by-coeff',repo/2.0775/r*cfc if False else repo/(2.0775*r))
# coefficient equality
for cfc in [60,180,600,1000]:
 b1,a1=sae_coeffs(cfc,fs); b2,a2=signal.butter(2,2.0775*cfc,fs=fs,btype='low')
 print(cfc,np.max(abs(b1-b2)),np.max(abs(a1-a2)))
```

```python
# Quantify filter transient decay and the effect of SciPy's 9-sample endpoint padding.
from scipy import signal
import numpy as np
fs=1.25e6
for cfc in [180,1000]:
    b,a=sae_coeffs(cfc,fs)
    # impulse response single pass; normalized tail envelope thresholds
    imp=np.zeros(int(fs*0.05)); imp[0]=1
    h=signal.lfilter(b,a,imp)
    peak=np.max(np.abs(h));
    for frac in [1e-2,1e-3,1e-4]:
        # last exceedance
        ix=np.where(np.abs(h)>peak*frac)[0]
        t=(ix[-1]/fs*1e3) if len(ix) else 0
        print(cfc,frac,t)
# Compare filtering a pulse in short record with default pad versus long reflected/constant quiet context
# not real data: deterministic algorithmic test signal, not empirical analysis
n=int(.020*fs); t=np.arange(n)/fs
x=np.zeros(n); event=.00035
# smooth half-sine 1 ms starting at event
idx=(t>=event)&(t<event+.001); x[idx]=np.sin(np.pi*(t[idx]-event)/.001)
b,a=sae_coeffs(180,fs)
ydef=signal.filtfilt(b,a,x)
for padlen_ms in [.01,.1,1,10]:
    pl=min(n-1,int(padlen_ms*1e-3*fs))
    yy=signal.filtfilt(b,a,x,padtype='odd',padlen=pl)
    print('pad ms',padlen_ms,'peak',yy.max(),'change%',100*(yy.max()/ydef.max()-1),'eventval',yy[np.argmax(idx)])
```
